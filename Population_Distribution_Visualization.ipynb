```python
## Problem Statement

To visualize the distribution of a categorical or continuous variable such as gender or age using bar charts and histograms.

```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
with open("Task_01_Dataset.csv", "r") as f:
    for i, line in enumerate(f):
        if i < 10:
            print(i, line)
```

```python
df = pd.read_csv(
    "Task_01_Dataset.csv",
    skiprows=4,      
    engine="python"
)

df.head()
```

```python
df.columns

```

```python
year = "2024"

population_2024 = df[[ "Country Name", year ]].dropna()
population_2024.head()
```

```python
plt.figure()
plt.hist(population_2024[year], bins=20)
plt.xlabel("Population")
plt.ylabel("Number of Countries")
plt.title("Population Distribution Across Countries (2024)")
plt.show()
```

```python
top10 = population_2024.sort_values(by=year, ascending=False).head(10)

plt.figure()
plt.bar(top10["Country Name"], top10[year])
plt.xticks(rotation=45, ha="right")
plt.xlabel("Country")
plt.ylabel("Population")
plt.title("Top 10 Most Populous Countries (2024)")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(population_2024["2024"], bins=30, edgecolor="black")
plt.xlabel("Population", fontsize=12)
plt.ylabel("Number of Countries", fontsize=12)
plt.title("Population Distribution Across Countries (2024)", fontsize=14)
plt.grid(axis="y", alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.bar(top10["Country Name"], top10["2024"])
plt.xticks(rotation=45, ha="right")
plt.xlabel("Country", fontsize=12)
plt.ylabel("Population", fontsize=12)
plt.title("Top 10 Most Populous Countries (2024)", fontsize=14)
plt.grid(axis="y", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
year_1="2020"
year_2="2024"
population_2000 = df[["Country Name", year_1]].dropna()

plt.figure(figsize=(10, 6))
plt.hist(population_2000[year_1], bins=30, edgecolor="black")
plt.xlabel("Population", fontsize=12)
plt.ylabel("Number of Countries", fontsize=12)
plt.title("Population Distribution Across Countries (2000)", fontsize=14)
plt.grid(axis="y", alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(population_2000[year_1], bins=30, alpha=0.6, label="2000")
plt.hist(population_2024[year_2], bins=30, alpha=0.6, label="2024")

plt.xlabel("Population", fontsize=12)
plt.ylabel("Number of Countries", fontsize=12)
plt.title("Population Distribution Comparison (2000 vs 2024)", fontsize=14)
plt.legend()
plt.grid(axis="y", alpha=0.6)
plt.show()
```

```python
top10_2000 = population_2000.sort_values(by=year_1, ascending=False).head(10)

plt.figure(figsize=(10, 6))
plt.bar(top10_2000["Country Name"], top10_2000[year_1])
plt.xticks(rotation=45, ha="right")
plt.xlabel("Country", fontsize=12)
plt.ylabel("Population", fontsize=12)
plt.title("Top 10 Most Populous Countries (2000)", fontsize=14)
plt.grid(axis="y", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
## Comparative Insights

- The overall population distribution has shifted towards higher values over time.
- High-population countries continue to dominate global population share.
- Comparing different years highlights long-term global population growth.
```
